```python
import pandas as pd
import sys
import os
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    RocCurveDisplay
)
import matplotlib.pyplot as plt
```

```python

PROJECT_ROOT = os.path.abspath("..")  
sys.path.insert(0, PROJECT_ROOT)

# mainīgie
#true_csv = os.path.join(PROJECT_ROOT, "data/labeled/labeled_df_leak_march.csv")
true_csv = os.path.join(PROJECT_ROOT, "test_outputs/anomaly_labeled_datasets/test1.csv")
pred_csv = os.path.join(PROJECT_ROOT, "test_outputs/anomaly_labeled_datasets/test16.csv")

true_df = pd.read_csv(true_csv)
pred_df = pd.read_csv(pred_csv)

#true_df = true_df.rename(columns={"anomaly": "over_threshold"})
```

```python
merged = true_df.merge(
    pred_df,
    on=["date", "delta"],
    suffixes=("_true", "_pred")
)

y_true = merged["over_threshold_true"].astype(int)
y_pred = merged["over_threshold_pred"].astype(int)

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, zero_division=0)
recall = recall_score(y_true, y_pred, zero_division=0)
f1 = f1_score(y_true, y_pred, zero_division=0)
```
